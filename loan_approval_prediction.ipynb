# Loan Approval Prediction

A machine learning project that predicts loan approval status using applicant financial features. The notebook uses Logistic Regression and Decision Tree classification.

## Features used
- ApplicantIncome
- CoapplicantIncome
- LoanAmount
- Credit_History

## Target
`Loan_Status` — `Y` (approved) or `N` (not approved).


```python
from google.colab import files
uploaded = files.upload()
import pandas as pd
import io
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score


loan_dataset = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[loan_dataset]))


print(f"Shape: {df.shape}")
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn names:")
print(df.columns.tolist())
print(df['Loan_Status'].value_counts())
df['Loan_Status'] = df['Loan_Status'].map({'Y': 1, 'N': 0})
print('---------------------------------------------------------------------------------------------------------')

print(df['Loan_Status'].value_counts())
feature_cols=['ApplicantIncome','CoapplicantIncome','LoanAmount','Credit_History']
target_col=['Loan_Status']
original=len(df)
df_clean=df[feature_cols+target_col]
df_clean=df_clean.dropna()
cleaned=len(df_clean)
print(f"original_rows:{original}")
print(f"after removing missing values:{cleaned}")
print(f"rows removed:{original-cleaned}")
print('---------------------------------------------------------------------------------------------------------')
x=df_clean[feature_cols]
y=df_clean['Loan_Status']
print(x.shape)
print(y.shape)
print('---------------------------------------------------------------------------------------------------------')
#training
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(f"Training data: {x_train.shape[0]} rows")
print(f"Testing data: {x_test.shape[0]} rows")
model=LogisticRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print('---------------------------------------------------------------------------------------------------------')
accuracy=model.score(x_test,y_test)
print(f"Accuracy: {accuracy*100:.2f}%")
print("Feature coefficients:")
print(f"ApplicantIncome: {model.coef_[0][0]:.4f}")
print(f"CoapplicantIncome: {model.coef_[0][1]:.4f}")
print(f"LoanAmount: {model.coef_[0][2]:.4f}")
print(f"Credit_History: {model.coef_[0][3]:.4f}")
print('---------------------------------------------------------------------------------------------------------')

#testing
test_uploaded = files.upload()

test_file = list(test_uploaded.keys())[0]
df_test = pd.read_csv(test_file)
x_test_new = df_test[feature_cols].fillna(0)
test_predictions = model.predict(x_test_new)
df_test['Predicted_Loan_Status'] = ['Y' if p == 1 else 'N' for p in test_predictions]


print("\nFirst 10 predictions:")
print(df_test[['Loan_ID', 'Predicted_Loan_Status']].head(10))

output_file = 'loan_approval_predictions.csv'
df_test[['Loan_ID', 'Predicted_Loan_Status']].to_csv(output_file, index=False)


```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
import io

# Get the uploaded file name
file_name = list(uploaded.keys())[0]

# Load the CSV
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

# Verify
print(df.head())
df['Loan_Status']= df['Loan_Status'].map({'Y':1,'N':0})
feature_cols=['ApplicantIncome','CoapplicantIncome','LoanAmount','Credit_History']
target_col=['Loan_Status']
df = df[feature_cols+ target_col].dropna()
x=df[feature_cols]
y=df['Loan_Status']
print(x.shape)
print(y.shape)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(f"Training set: {x_train.shape[0]} rows")
print(f"Test set: {x_test.shape[0]} rows")
model=DecisionTreeClassifier(random_state=42)
model.fit(x_train,y_train)
print(f"Tree depth: {model.get_depth()}")
print(f"Number of leaves: {model.get_n_leaves()}")
print("Feature importance:")
for col, imp in zip(feature_cols, model.feature_importances_):
    print(f"  {col}: {imp:.4f}")
```
